```python
#Trapezoid Rule
def f(x):
    return ((x**2-1)**1/2)

a = 3   #lower limit
b = 7   #upper
n = 5   #a theke b porjonto 6ta trapezoid
del_x = (b-a)/n
sum = 0
for i in range(1,n):
  sum += f(a + i*del_x)

value_of_area = del_x/2*(f(a)+2*sum+f(b))
print(value_of_area)
```

```python
#differential equation
import sympy as sp
x = sp.symbols("x")

y = sp.Function("y")(x)  #function of x
expr = sp.Eq(sp.diff(y,x),2*x)
display(expr)
solve1 = sp.dsolve(expr)
display(solve1)
```

```python
#initial conditon
#boundary condition
#to get the value of and get the exact one and only function
#differential equation
import sympy as sp
x = sp.symbols("x")

y = sp.Function("y")(x)  #function of x
expr = sp.Eq(sp.diff(y,x),2*x)

boundary = {y.subs(x,0):2}

solve1 = sp.dsolve(expr,ics=boundary)
display(solve1)
```

```python
x = sp.symbols("x") #x independent variable
f = sp.Function("f")(x)

expr = sp.Eq(x*sp.diff(f,x,2) + sp.diff(f,x),x**3)

boundary = {f.subs(x,1):0,sp.diff(f,x,2).subs(x,2):1} #boundary value 2ta lagbe

result = sp.dsolve(expr,ics = boundary)
display(result) #2ta c ashe
```

```python
#graph e boshai
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np
x = sp.symbols("x") #x independent variable
f = sp.Function("f")(x)

expr = sp.Eq(x*sp.diff(f,x,2) + sp.diff(f,x),x**3)

boundary = {f.subs(x,1):0,sp.diff(f,x,2).subs(x,2):1} #boundary value 2ta lagbe

result = sp.dsolve(expr,ics = boundary)
display(result)
x_val = np.linspace(0,2,100) #array banay  #100ta point dibe
y_val = []   # Create an empty list to store y values
#ekhon kaj y er value ber kora
#x er value function e boshale
# y er o value peye jabo

for val in x_val:
    y_val.append(result.rhs.evalf(subs={x: val}))  # Append each y value to the list
print(y_val)  #evalf numeric value dey

plt.plot(x_val,y_val)
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

f = lambda x, y: np.exp(x) * np.sin(x + y)  # dy/dx = e^x*sin(x+y)
a = 0
b = 5
n = 1001
h = (b - a) / (n - 1)

x = np.linspace(a, b, n)
y = np.zeros_like(x)
y[0] = 1

for i in range(1, n):
    y[i] = y[i - 1] + h * f(x[i - 1], y[i - 1])

plt.plot(x, y)
plt.grid(True)
plt.show()
```

1.solve the differential equation where take b =1 and g = 10
2.plot v vs t
3.Plot v vs t in different values of b,i.e b = 1 , b = 2 ,b =3,b =-4and b=2

```python
import numpy as np
import matplotlib.pyplot as plt

# Function to define the ODE
def ode(v, t, b, g):
    return g - b * v**2

# Function to solve the ODE using Euler's method
def solve_ode(b, g, dt, T):
    t_values = np.arange(0, T, dt)
    v_values = [0]  # Initial velocity is 0

    for t in t_values[:-1]:
        v_next = v_values[-1] + ode(v_values[-1], t, b, g) * dt
        v_values.append(v_next)

    return t_values, v_values

# Parameters
b = 1
g = 10
dt = 0.01
T = 5

# Solve the ODE for b=1
t_values, v_values_b1 = solve_ode(b, g, dt, T)

# Plot v vs t for b=1
plt.figure(figsize=(8, 6))
plt.plot(t_values, v_values_b1, label='b=1')
plt.title('Freely Falling Object with Air Resistance (b=1)')
plt.xlabel('Time (t)')
plt.ylabel('Velocity (v)')
plt.legend()
plt.show()
```
